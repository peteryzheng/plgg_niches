```python
def geojson_contains_xy(geojson, xy):
    hits = geojson[geojson.contains(xy)]
    if hits.empty:
        return {"annotation": "unannotated", "region_id": pd.NA}

    hit = hits.iloc[0]
    region_id = hit.get("id")
    if pd.isna(region_id):
        region_id = hit.name
    return {"annotation": hit["color"], "region_id": str(region_id)}
```
